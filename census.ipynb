```python
df['Sex_ratio']=(df['Female']/df['Male'])*1000
state_sex_ratio = df.groupby('State name')['Sex_ratio'].mean().sort_values(ascending=False)
print(state_sex_ratio)
```

```python
state_sex_ratio.head(10).plot(kind='bar',figsize=(10,6),color='red')
plt.title('Top 10 states of india listed by sexratio')
plt.xlabel('State name')
plt.ylabel('Sex ratio')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

x=df[['Population','Male','Female']]
y=df['Literate']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
y_pred[:10]
```

```python
from sklearn.metrics import mean_squared_error, r2_score

# Calculate error (how far off predictions are)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("R² Score:", r2)
```

```python
import matplotlib.pyplot as plt

plt.scatter(y_test,y_pred,alpha=0.6,color='blue')
plt.xlabel('Actual Literate')
plt.ylabel('Predicted Literate')
plt.title('Actual vs Predicted Literate')
plt.show()
```
